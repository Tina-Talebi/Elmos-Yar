```python
def extract_text(t):
    if isinstance(t, list):
        parts = []
        for item in t:
            if isinstance(item, dict) and "text" in item:
                parts.append(str(item["text"]))
            else:
                parts.append(str(item))
        return "".join(parts)
    return str(t)
```

```python
 

def parse_message(msg):
    # Professor
    professor_name = re.search(r'🧑‍🏫\s*(.*)', msg)
    professor_name = professor_name.group(1).strip() if professor_name else None

    # Department
    department = re.search(r'🏫\s*(#\S+)', msg)
    department = department.group(1).strip() if department else None

    # Course
    course_name = re.search(r'📒\s*(.*)', msg)
    course_name = course_name.group(1).strip() if course_name else None

    # Attendance
    attendance = re.search(r'حضور و غیاب\s*\n\s*┘\s*(.*)', msg)
    attendance = attendance.group(1).strip() if attendance else None

    # Grading
    grading = re.search(r'وضعیت نمره دادن:\s*\n\s*┘\s*(.*)', msg)
    grading = grading.group(1).strip() if grading else None

    # Ratings (from ارزیابی دانشجو)
    ratings = re.findall(r':\s*(\d+)', msg)
    ratings = [int(r) for r in ratings]
    while len(ratings) < 6:
        ratings.append(None)
    rating_1, rating_2, rating_3, rating_4, rating_5, rating_6 = ratings[:6]

    # Term
    term = re.search(r'ترمی که دانشجو.*?\n\s*┘\s*(.*)', msg)
    term = term.group(1).strip() if term else None

    # Comment (توضیحات)
    comment = re.search(r'توضیحات:\s*\n\s*┘\s*(.*)', msg, re.DOTALL)
    comment = comment.group(1).strip() if comment else None

    return {
        'professor_name_raw': professor_name,
        'department': department,
        'course_name': course_name,
        'comment_text': comment,
        'grading_status_raw': grading,
        'attendance_status_raw': attendance,
        'rating_1': rating_1,
        'rating_2': rating_2,
        'rating_3': rating_3,
        'rating_4': rating_4,
        'rating_5': rating_5,
        'rating_6': rating_6,
        'term': term
    }
```

```python
import re
import pandas as pd

def extract_review_fields(review_text):
    # --- Input Validation ---
    if not isinstance(review_text, str):
        return dict(
            sources=None, attendance=None, exam_sufficient=None, grading=None,
            rating_1=None, rating_2=None, rating_3=None, rating_4=None, rating_5=None, rating_6=None,
            contact=None, term=None, comments=None
        )

    # Use re.DOTALL flag for all multiline searches

    # --- 1. Sources ---
    # Looks for 'منابع آموزش' followed by optional spaces, newline, ┘, optional spaces, and then captures
    sources_match = re.search(r'منابع آموزش\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    sources = sources_match.group(1).strip() if sources_match else None

    # --- 2. Attendance ---
    attendance_match = re.search(r'حضور و غیاب\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    attendance = attendance_match.group(1).strip() if attendance_match else None

    # --- 3. Exam sufficient ---
    exam_match = re.search(r'منابع معرفی شده برای امتحان کافی است\؟\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    exam_sufficient = exam_match.group(1).strip() if exam_match else None

    # --- 4. Grading ---
    grading_match = re.search(r'وضعیت نمره دادن:\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    grading = grading_match.group(1).strip() if grading_match else None

    # --- 5. Ratings (This part was mostly correct) ---
    # Finds lines starting with '┤', followed by any text and a colon, then captures the digit.
    # The 'rating_matches' should be sufficient if the order is always the same.
    rating_matches = re.findall(r'┤ .*?:\s*(\d+)', review_text)
    ratings = [int(r) for r in rating_matches[:6]]
    
    # Pad with None if fewer than 6 ratings were found
    while len(ratings) < 6:
        ratings.append(None)
    
    r1, r2, r3, r4, r5, r6 = ratings

    # --- 6. Contact ---
    contact_match = re.search(r'راه ارتباطی:\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    contact = contact_match.group(1).strip() if contact_match else None

    # --- 7. Term ---
    term_match = re.search(r'ترمی که دانشجو با این استاد کلاس داشته:\s*\n\s*┘\s*(.*?)\n', review_text, re.DOTALL)
    term = term_match.group(1).strip() if term_match else None

    # --- 8. Comments (Crucial Fix) ---
    # Looks for 'توضیحات:' followed by optional spaces, newline, ┘, optional spaces, then captures
    # everything (.*?) up until the tilde pattern (~{10,}) or the end of the string ($).
    comments_match = re.search(r'توضیحات:\s*\n\s*┘\s*(.*?)(?:~{10,}|$)[\s\S]*$', review_text, re.DOTALL)
    comments = comments_match.group(1).strip() if comments_match else None

    return dict(
        sources=sources, attendance=attendance, exam_sufficient=exam_sufficient, grading=grading,
        rating_1=r1, rating_2=r2, rating_3=r3, rating_4=r4, rating_5=r5, rating_6=r6,
        contact=contact, term=term, comments=comments
    )

# --- Example Usage on Your Sample Text ---
sample_text = """
🧑‍🏫 فاطمه مهدی زاده سراج
🏫 #معماری_و_شهرسازی
📒 مبانی طراحی ۲

منابع آموزش
  ┘ منبعی معرفی نمی کنند

حضور و غیاب
  ┘ یادم نمیاد

منابع معرفی شده برای امتحان کافی است؟
  ┘ بله

وضعیت نمره دادن:
  ┘ منصفانه و هرچی خودت بگیری

ارزیابی دانشجو:
  ┤ پیوستگی و یکپارچگی تدریس: 10
  ┤ دانش عمومی: 10
  ┤ توانایی انتقال مطالب و نکات مهم درس: 10
  ┤ نحوه مدیریت کلاس(نظم و زمان): 10
  ┤ پاسخگویی(حضوری و غیرحضوری): 10
  ┤ آداب و رفتار اجتماعی با دانشجویان: 10

راه ارتباطی:
  ┘ چیزی ندارم

ترمی که دانشجو با این استاد کلاس داشته:
  ┘ بهمن ۱۴۰۳

توضیحات:
  ┘ با دکتر طباطبایی و دو سه تا تی ای میان سر کلاس
خوبه کلاساش استفاده کنین چیزای خوبی یاد میگیرین
عکس و اینام بگیرید از کاراتون در طول ترم آخر ترم جمع نشه یه جا
ایده هم خیلی مهمه براشون
~~~~~~~~~~~~~~~~~
برای ثبت معرفی استاد به ربات زیر پیام بدید
@ostad_elmosiBot

کانال معرفی اساتید دانشگاه علم و صنعت
@ostad_elmosi @ostad_elmosi
"""

extracted_data = extract_review_fields(sample_text)
print(extracted_data)
```

```python
import re
 

def remove_emojis(text):
    if text is None:
        return None
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
        u"\U00002702-\U000027B0"
        u"\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text).strip()

def extract_full_review(review_text):
    """
    Extracts ALL structured fields from raw review text, with a final, highly 
    robust fix for the department name extraction.
    """
    if not isinstance(review_text, str):
        return dict(
            professor_name_raw=None, department=None, course_name=None,
            sources=None, attendance=None, exam_sufficient=None, grading=None,
            rating_1=None, rating_2=None, rating_3=None, rating_4=None, rating_5=None, rating_6=None,
            contact=None, term=None, comments=None
        )

    # Use re.DOTALL for multiline matches in text blocks
    
    # --- 1. Extract Identifying Fields (Focus on Department Fix) ---
    
    # Professor Name: Capture everything after 🧑‍🏫 up until the next line break or 🏫
    prof_match = re.search(r'🧑‍🏫\s*(.*?)(?=\n\s*🏫|\n\s*\S*#|\n|\Z)', review_text, re.DOTALL)
    professor_name_raw = prof_match.group(1).strip() if prof_match else None
    
    # 🏫 DEPARTMENT FIX 🛠️: Look specifically for the department name after 🏫 or #,
    # ensuring it stops before the course name starts.
    department = None
    
    # 1. Try finding department text after 🏫 (Emoji header)
    dept_match_emoji = re.search(r'🏫\s*(.*?)(?=\s*📒|\s*منابع آموزش|\Z)', review_text, re.DOTALL)
    
    # 2. Try finding department text after # (Hashtag format, often cleaner)
    # This pattern looks for the first # and captures until 📒, Sources, or the end.
    dept_match_hash = re.search(r'#\s*(.*?)(?=\s*📒|\s*منابع آموزش|\Z)', review_text, re.DOTALL)
    
    # Prioritize the Hash match if it exists and contains more text than the Emoji match (to handle professor name duplication)
    if dept_match_hash:
        department = remove_emojis(dept_match_hash.group(1).strip())
    elif dept_match_emoji:
        # If no hash is found, use the emoji match.
        department = remove_emojis(dept_match_emoji.group(1).strip())
    
    # Final cleanup: If the department still contains the professor's name (common failure case),
    # and the professor's name is present, try to remove the professor's name and any
    # trailing identical text from the department field.
    if professor_name_raw and department:
        if department.startswith(professor_name_raw):
            department = department.replace(professor_name_raw, '').strip()
        # Clean up any remaining leading/trailing characters like - or |
        department = re.sub(r'^[-\s|]+|[-\s|]+$', '', department)
        
    # 📒 Course Name: Capture text after 📒 up until the Sources header
    course_match = re.search(r'📒\s*(.*?)(?=\s*منابع آموزش|\Z)', review_text, re.DOTALL)
    course_name = remove_emojis(course_match.group(1).strip()) if course_match else None


    # --- 2. Extract Body Fields (No change needed, the previous version was good) ---
    sources_match = re.search(r'منابع آموزش\s*\n?\s*┘\s*(.*?)(?=\s*حضور و غیاب|\Z)', review_text, re.DOTALL)
    sources = remove_emojis(sources_match.group(1).strip()) if sources_match else None

    attendance_match = re.search(r'حضور و غیاب\s*\n?\s*┘\s*(.*?)(?=\s*منابع معرفی شده برای امتحان|\Z)', review_text, re.DOTALL)
    attendance = remove_emojis(attendance_match.group(1).strip()) if attendance_match else None

    exam_match = re.search(r'منابع معرفی شده برای امتحان کافی است\؟\s*\n?\s*┘\s*(.*?)(?=\s*وضعیت نمره دادن|\Z)', review_text, re.DOTALL)
    exam_sufficient = remove_emojis(exam_match.group(1).strip()) if exam_match else None

    grading_match = re.search(r'وضعیت نمره دادن:\s*\n?\s*┘\s*(.*?)(?=\s*ارزیابی دانشجو|\Z)', review_text, re.DOTALL)
    grading = remove_emojis(grading_match.group(1).strip()) if grading_match else None

    contact_match = re.search(r'راه ارتباطی:\s*\n?\s*┘\s*(.*?)(?=\s*ترمی که دانشجو با این استاد کلاس داشته:|\Z)', review_text, re.DOTALL)
    contact = remove_emojis(contact_match.group(1).strip()) if contact_match else None

    term_match = re.search(r'ترمی که دانشجو با این استاد کلاس داشته:\s*\n?\s*┘\s*(.*?)(?=\s*توضیحات:|\Z)', review_text, re.DOTALL)
    term = remove_emojis(term_match.group(1).strip()) if term_match else None


    # --- 3. Extract Ratings (Numerical Fields) ---
    rating_matches = re.findall(r'┤ .*?:\s*(\d+(?:\.\d+)?)', review_text)
    ratings = [float(r) if r else None for r in rating_matches[:6]]
    
    while len(ratings) < 6:
        ratings.append(None)
    
    r1, r2, r3, r4, r5, r6 = ratings

    # --- 4. Extract Comments (Final field) ---
    comments_match = re.search(r'توضیحات:\s*\n?\s*┘\s*(.*?)(?:~{2,}|$)[\s\S]*$', review_text, re.DOTALL)
    comments = remove_emojis(comments_match.group(1).strip()) if comments_match else None

    return dict(
        professor_name_raw=professor_name_raw, department=department, course_name=course_name,
        sources=sources, attendance=attendance, exam_sufficient=exam_sufficient, grading=grading,
        rating_1=r1, rating_2=r2, rating_3=r3, rating_4=r4, rating_5=r5, rating_6=r6,
        contact=contact, term=term, comments=comments
    )
```
