```python
def extract_text(t):
    if isinstance(t, list):
        parts = []
        for item in t:
            if isinstance(item, dict) and "text" in item:
                parts.append(str(item["text"]))
            else:
                parts.append(str(item))
        return "".join(parts)
    return str(t)
```

```python
# Extract scores
def extract_consistency_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'پیوستگی و یکپارچگی تدریس:\s*(\d+)', msg)
    return m if m else None

def extract_knowledge_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'دانش عمومی:\s*(\d+)', msg)
    return m if m else None

def extract_delivery_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'توانایی انتقال مطالب و نکات مهم درس:\s*(\d+)', msg)
    return m if m else None

def extract_management_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'نحوه مدیریت کلاس\(نظم و زمان\):\s*(\d+)', msg)
    return m if m else None

def extract_responsiveness_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'پاسخگویی\(حضوری و غیرحضوری\):\s*(\d+)', msg)
    return m if m else None

def extract_respect_score(msg):
    # Find all numeric ratings in ارزیابی دانشجو section
    m = re.findall(r'آداب و رفتار اجتماعی با دانشجویان:\s*(\d+)', msg)
    return m if m else None
```

### task 3: یکی کردن اسامی اساتید

```python
titles = [
    "استاد",
    "دکتر",
    "مهندس",
    "سید",
    "سیده",
    "خانم",
    "آقای",
    "پروفسور"
]

def remove_titles(name):
    for t in titles:
        name = re.sub(rf"^{t}\s+", "", name)
    return name.strip()
```

```python
def normalize_persian(text):
    replacements = {
        "ي": "ی",
        "ك": "ک"
        }
    
    for k, v in replacements.items():
        text = text.replace(k, v)

    # collapse multiple spaces
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
df_reviews["professor_name"] = df_reviews["professor_name"].apply(normalize_persian)
df_reviews["professor_name"] = df_reviews["professor_name"].apply(remove_titles)
```

### task 4: یکی کردن اسم درس و برچسب های متنی

```python
# --- 1) translation tables ---
EN2FA_DIGITS = str.maketrans("0123456789", "۰۱۲۳۴۵۶۷۸۹")

AR2FA_LETTERS = str.maketrans({
    "ك": "ک",
    "ي": "ی",
    "ى": "ی"
})

ZWNJ = "\u200c"  # half-space
FA_DIGITS = "۰۱۲۳۴۵۶۷۸۹"
DIGIT_CLASS = f"[{FA_DIGITS}]"
```

```python
# --- 2) main normalizer exactly per your spec ---
def clean_course_text(x):
    if pd.isna(x):
        return ""
    s = str(x)

    # english digits -> persian digits
    s = s.translate(EN2FA_DIGITS)

    # arabic letters -> persian letters (ک, ی)
    s = s.translate(AR2FA_LETTERS)

    # half-space -> space
    s = s.replace(ZWNJ, " ")

    # normalize whitespace
    s = re.sub(r"\s+", " ", s).strip()

    # word to number
    s = s.replace(' یک', ' ۱')
    s = s.replace(' دو', ' ۲')
    s = s.replace(' سه', ' ۳')

    # IMPORTANT: keep '-' and '_' for splitting later,
    # but first fix cases like "ریاضی - ۲" where dash is just formatting before a number:
    # remove any '-' right before a digit (with any spaces around)
    s = re.sub(rf"\s*-\s*(?={DIGIT_CLASS})", " ", s)

    # now enforce: if a digit appears and previous char isn't space or a digit, add exactly one space before it
    # e.g. "ریاضی۲" -> "ریاضی ۲"
    s = re.sub(rf"(?<![\s{FA_DIGITS}])({DIGIT_CLASS})", r" \1", s)

    # collapse multiple spaces again
    s = re.sub(r"\s+", " ", s).strip()

    s = (c := s.replace("عمومی", "").strip()) or s
    s = s.replace("انگلیسی", "").strip()
    s = s.replace("ریاضیات", "ریاضی")
    

    return s
```

```python
SPLIT_RE = re.compile(r"\s*[-_]\s*")

def split_on_dash_underscore(s):
    s = clean_course_text(s)
    if not s:
        return []
    parts = [p.strip() for p in SPLIT_RE.split(s) if p and p.strip()]
    return parts
```

```python
# Step 1: Add normalized column
df_reviews["course_normal"] = df_reviews["course"].apply(clean_course_text)

# Step 2: Split into list (temporary column)
df_reviews["_course_split"] = df_reviews["course_normal"].map(split_on_dash_underscore)

# Step 3: Explode the temporary column to create final column
df_reviews["course_clean"] = df_reviews["_course_split"]  # Copy first
df_reviews = df_reviews.explode("course_clean")

# Step 4: Remove temporary column
df_reviews = df_reviews.drop(columns=["_course_split"])

# Step 5: Reset index
df_reviews = df_reviews.reset_index(drop=True)
```
