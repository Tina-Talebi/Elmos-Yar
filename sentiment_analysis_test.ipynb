```python
import pandas as pd
from transformers import pipeline
import hazm  # For Persian text normalization

# Load your CSV (use your full_cleaned_reviews.csv; for testing, assume attached is similar)
df = pd.read_csv('full_cleaned_reviews.csv')  # Or 'extracted_reviews_v3_final.csv' for the attached data

# Preprocess function (normalize Persian text for better accuracy)
normalizer = hazm.Normalizer()
def preprocess_comment(comment):
    if pd.isna(comment) or not comment.strip():
        return ""
    return normalizer.normalize(comment)  # Fixes Arabic/Persian chars, removes extras

df['comments'] = df['comments'].apply(preprocess_comment)

# Load ternary Persian sentiment model
sentiment_pipeline = pipeline("sentiment-analysis", model="Behpouyan/Behpouyan-Sentiment")

# Function to tag and score
def analyze_sentiment(comment):
    if not comment:
        return "neutral", 0.0
    result = sentiment_pipeline(comment)[0]
    label = result['label'].lower()  # 'positive', 'negative', 'neutral'
    # Score: confidence * polarity (-1 to +1)
    polarity = 1 if label == 'positive' else -1 if label == 'negative' else 0
    score = result['score'] * polarity
    # Optional: If confidence low (<0.6), force neutral for edge cases
    if result['score'] < 0.6:
        return "neutral", 0.0
    return label, score

# Apply to DataFrame
df[['sentiment_tag', 'sentiment_score']] = df['comments'].apply(analyze_sentiment).apply(pd.Series)

# Group by professor: Count tags (amount) and avg score (tone)
summary = df.groupby('professor_name_raw').agg({
    'sentiment_tag': lambda x: x.value_counts().to_dict(),  # e.g., {'positive': 3, 'negative': 1, 'neutral': 2}
    'sentiment_score': 'mean'  # Avg tone (-1 to +1)
}).reset_index()

print("Summary (first 5):")
print(summary.head())

# Save for app (use in Streamlit for profiles/recommender)
df.to_csv('reviews_with_sentiment.csv', index=False)
print("Saved to reviews_with_sentiment.csv")
```

```python
import pandas as pd
import re # Make sure 're' is imported for the preprocessing function
from hazm import Normalizer, word_tokenize, stopwords_list
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

# Load your CSV
# NOTE: Ensure this path is correct
df = pd.read_csv(r'D:\ElMosYar\ElMosYar\extracted_reviews_v3_final.csv')

# Preprocessing (your same one)
normalizer = Normalizer()
stopwords = stopwords_list()

def preprocess(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = normalizer.normalize(text)
    text = re.sub(r'[^\w\s]', '', text)
    tokens = [t for t in word_tokenize(text) if t not in stopwords]
    return " ".join(tokens)

df['cleaned_comments'] = df['comments'].apply(preprocess)

# MODEL LOADING (Unchanged - this part is correct)
model_name = "HooshvareLab/bert-fa-base-uncased-sentiment-snappfood"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

pipe = pipeline(
    "text-classification",
    model=model,
    tokenizer=tokenizer,
    device=-1,        # ← CPU
    batch_size=64
)

# --- THE FIX IS HERE ---
def get_sentiment(text):
    if not text.strip():
        return 0.0
    try:
        res = pipe(text[:512])[0]
        label = res['label']
        score = res['score']
        
        # CORRECTED LOGIC: Check for the model's actual positive label ('HAPPY')
        # HAPPY -> Positive Score
        # SAD -> Negative Score
        return round(score, 4) if label == "HAPPY" else round(-score, 4)
    except Exception as e:
        # Catch any pipeline errors (e.g., excessive length, CUDA memory issues, etc.)
        print(f"Error processing text: {text[:30]}... Error: {e}")
        return 0.0

print("Running (will finish in ≈70–90 seconds)...")
df['sentiment_score'] = df['cleaned_comments'].apply(get_sentiment)

# Use thresholds to map score to final label
df['sentiment_label'] = df['sentiment_score'].apply(
    lambda x: "Positive" if x > 0.3 else ("Negative" if x < -0.3 else "Neutral")
)

# Save
df.to_csv('professor_sentiment_snappfood_model_FIXED.csv', index=False, encoding='utf-8-sig')
print("Done! First 15 rows (with corrected scores):")
print(df[['comments', 'sentiment_score', 'sentiment_label']].head(15))
```

```python
# ===============================================
# FINAL SENTIMENT PIPELINE - PERFECTED RULES
# ===============================================
import pandas as pd
import numpy as np
import random
import re
from collections import Counter
import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
import torch.nn.functional as F
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import RandomOverSampler
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup

# SEEDS
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True

# CONFIG
CSV_PATH = "full_cleaned_reviews.csv"
MODEL_NAME = "HooshvareLab/bert-fa-base-uncased"
MAX_LEN, BATCH_SIZE = 128, 16
NUM_EPOCHS, PATIENCE = 6, 3
LEARNING_RATE = 3e-5

label2id = {"negative": 0, "neutral": 1, "positive": 2}
id2label = {v: k for k, v in label2id.items()}

print("✅ Imports ready")

# ===============================================
# 1. CLEANING FUNCTION
# ===============================================
def clean_comment(text: str) -> str:
    text = str(text).strip()
    text = re.sub(r"http\S+|www\S+|@\S+|\d+", "", text)
    text = re.sub(r"[^\w\s]", " ", text)
    return " ".join(text.split())

# ===============================================
# 2. LOAD & PREPARE DATA
# ===============================================
df = pd.read_csv(CSV_PATH, encoding="utf-8")
df = df.dropna(subset=["comments"])
df = df[df["comments"].str.strip() != ""]
df["comments"] = df["comments"].apply(clean_comment)

rating_cols = ["rating_1", "rating_2", "rating_3", "rating_4", "rating_5", "rating_6"]
for c in rating_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df["rating_mean"] = df[rating_cols].mean(axis=1, skipna=True)

def rating_to_sentiment(x: float):
    if pd.isna(x): return None
    if x <= 5.0: return "negative"
    elif x < 8.0: return "neutral"
    return "positive"

df["sentiment"] = df["rating_mean"].apply(rating_to_sentiment)
df = df.dropna(subset=["sentiment"])

print("Labels:", df["sentiment"].value_counts().to_dict())
data = df[["comments", "sentiment"]].reset_index(drop=True)

# ===============================================
# 3. TRAIN/VAL SPLIT + OVERSAMPLE
# ===============================================
train_df, val_df = train_test_split(data, test_size=0.2, random_state=SEED, stratify=data["sentiment"])
ros = RandomOverSampler(random_state=SEED)
train_texts, train_labels = ros.fit_resample(train_df["comments"].values.reshape(-1, 1), train_df["sentiment"])
train_df = pd.DataFrame({"comments": train_texts.flatten(), "sentiment": train_labels})

print(f"Train: {len(train_df)} | Val: {len(val_df)}")

# ===============================================
# 4. MODEL & DATASETS
# ===============================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

class ReviewsDataset(Dataset):
    def __init__(self, texts, labels):
        self.texts = list(texts)
        self.labels = [label2id[l] for l in labels]
    def __len__(self): return len(self.texts)
    def __getitem__(self, idx):
        enc = tokenizer(str(self.texts[idx]), truncation=True, padding="max_length", max_length=MAX_LEN, return_tensors="pt")
        return {k: v.squeeze(0) for k, v in enc.items()} | {"labels": torch.tensor(self.labels[idx])}

train_dataset = ReviewsDataset(train_df["comments"], train_df["sentiment"])
val_dataset = ReviewsDataset(val_df["comments"], val_df["sentiment"])
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=3, id2label=id2label, label2id=label2id).to(device)

# ===============================================
# 5. TRAINING
# ===============================================
class_weights = compute_class_weight("balanced", classes=np.unique(train_df["sentiment"]), y=train_df["sentiment"])
class_weights = torch.tensor(class_weights, dtype=torch.float).to(device)
loss_fn = torch.nn.CrossEntropyLoss(weight=class_weights)
optimizer = AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=0.01)

total_steps = len(train_loader) * NUM_EPOCHS
scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps=int(0.1 * total_steps), num_training_steps=total_steps)

def train_epoch():
    model.train(); total_loss = 0
    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = model(input_ids=batch["input_ids"], attention_mask=batch["attention_mask"])
        loss = loss_fn(outputs.logits, batch["labels"])
        optimizer.zero_grad(); loss.backward(); optimizer.step(); scheduler.step()
        total_loss += loss.item()
    return total_loss / len(train_loader)

def eval_epoch():
    model.eval(); preds, trues, total_loss = [], [], 0
    with torch.no_grad():
        for batch in val_loader:
            batch = {k: v.to(device) for k, v in batch.items()}
            outputs = model(input_ids=batch["input_ids"], attention_mask=batch["attention_mask"])
            loss = loss_fn(outputs.logits, batch["labels"])
            total_loss += loss.item()
            preds.extend(torch.argmax(outputs.logits, -1).cpu().numpy())
            trues.extend(batch["labels"].cpu().numpy())
    report = classification_report(trues, preds, target_names=["negative", "neutral", "positive"], digits=3, zero_division=0)
    return total_loss / len(val_loader), report

# EARLY STOPPING
best_val_loss = float('inf'); best_state = None; patience_counter = 0
for epoch in range(NUM_EPOCHS):
    train_loss = train_epoch()
    val_loss, report = eval_epoch()
    print(f"\nEpoch {epoch+1}/{NUM_EPOCHS} | Train: {train_loss:.4f} | Val: {val_loss:.4f}")
    print(report)

    if val_loss < best_val_loss - 1e-3:  # Stricter threshold
        best_val_loss = val_loss; best_state = model.state_dict()
        patience_counter = 0; print("✅ Best!")
    else:
        patience_counter += 1
        print(f"Patience: {patience_counter}/{PATIENCE}")
        if patience_counter >= PATIENCE: break

model.load_state_dict(best_state)
model.save_pretrained("bert_professor_sentiment_model")
tokenizer.save_pretrained("bert_professor_sentiment_model")

# ===============================================
# 6. FINAL PREDICTIONS
# ===============================================
model.eval()
def predict_all(texts):
    preds, confs = [], []
    for i in range(0, len(texts), BATCH_SIZE):
        batch = tokenizer(list(texts[i:i+BATCH_SIZE]), truncation=True, padding=True, max_length=MAX_LEN, return_tensors="pt").to(device)
        with torch.no_grad():
            outputs = model(**batch)
            probs = F.softmax(outputs.logits, -1).cpu().numpy()
            preds.extend([id2label[np.argmax(p)] for p in probs])
            confs.extend([np.max(p) for p in probs])
    return preds, confs

# FULL DATASET
full_df = pd.read_csv(CSV_PATH, encoding="utf-8")
full_df = full_df.dropna(subset=["comments"])
full_df["comments"] = full_df["comments"].apply(clean_comment)

full_df["model_sentiment"], full_df["confidence"] = predict_all(full_df["comments"])

# RATINGS
for c in rating_cols:
    full_df[c] = pd.to_numeric(full_df[c], errors="coerce")
full_df["rating_mean"] = full_df[rating_cols].mean(axis=1, skipna=True)
full_df["rating_sentiment"] = full_df["rating_mean"].apply(rating_to_sentiment)

# ===============================================
# 7. FIXED HYBRID RULES (NO LENGTH LIMIT!)
# ===============================================
GENERIC_PHRASES = ["چیزی اضافه", "چیزی ندارم", "چیزی نیست", "نمی دونم", "نمیگم"]

def final_sentiment(row):
    text = str(row["comments"])

    # 1. TRUE GENERIC PHRASES → neutral (NOT short text!)
    text_lower = text.lower()
    if any(phrase in text_lower for phrase in GENERIC_PHRASES):
        return "neutral"

    # 2. Model + rating agreement → trust
    if row["model_sentiment"] == row["rating_sentiment"]:
        return row["model_sentiment"]

    # 3. Default: trust model
    return row["model_sentiment"]

full_df["final_sentiment"] = full_df.apply(final_sentiment, axis=1)

# SAVE
full_df.to_csv("FINAL_reviews_sentiment_analysis.csv", index=False, encoding="utf-8")
print("\n🎉 COMPLETE!")
print(full_df[["comments", "rating_sentiment", "model_sentiment", "final_sentiment", "confidence"]].head(15))
print("\nDistribution:", full_df["final_sentiment"].value_counts().to_dict())
```
